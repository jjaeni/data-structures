## 2. 만든거 테스트

```python
from DequeueAlgorithm import Dequeue
```

```python
s = input()
dq = Dequeue(s)
palindrom = True

while len(dq)>1:
    dq.popleft()
    print(len(dq))
```

```python
def check_palindrom(s):
    dq = Dequeue(s)
    palindrom = True

    while len(dq)>1:
        if dq.popleft() != dq.pop():
            palindrom = False
    return palindrom

s = input('문자열 입력해')
print(check_palindrom(s))
```
