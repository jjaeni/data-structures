## 2. 초안

```python
opstack = Stack()
outstack = []
token_list = "( A + B ) * C".split()

print(f'token list: {token_list}')

prec = {}
prec['('] = 0
prec['+'] = 3
prec['-'] = 3
prec['*'] = 2
prec['/'] = 2
prec['^'] = 1

print(f'prec: {prec} \n')

for token in token_list:
    if token == '(':
        opstack.push(token)
    elif token == ')':
        while opstack.top() != '(':
            outstack.append(opstack.pop())
        opstack.push(token)
    elif token in '+-/*^':
        filtered_prec = {k:v for k, v in prec.items() if v <= prec[token] < 0}
        if opstack.top() in filtered_prec.keys():
            outstack.append(opstack.pop())
        opstack.push(token)
    else:
        outstack.append(token)
    print(f'token: {token} \noutstack: {outstack} \n')
    
while opstack.top() != ')':
    outstack.append(opstack.pop())

print(outstack)
```

## 3. 테스트 중

```python
'''
Infix to postfix
'''


class Stack:
    def __init__(self):
        self.items = []

    def push(self, val):
        self.items.append(val)

    def pop(self):
        try:
            return self.items.pop()
        except IndexError:
            print("Stack is empty")

    def top(self):
        try:
            return self.items[-1]
        except IndexError:
            print("Stack is empty")

    def __len__(self):
        return len(self.items)

    def isEmpty(self):
        return self.__len__() == 0

def infix_to_postfix(infix):
    opstack = Stack()
    outstack = []
    token_list = infix.split()
    print(token_list)
		
    # 연산자의 우선순위 설정
    prec = {}
    prec['('] = 0
    prec['+'] = 3
    prec['-'] = 3
    prec['*'] = 2
    prec['/'] = 2
    prec['^'] = 1

    for token in token_list:
        if token == '(':
            opstack.push(token)
            print(token, 'opstack에 push')

        elif token == ')':
            while opstack.top() != '(':
                outstack.append(opstack.pop())
                print('기존 opstack의 연산자를 outstack에 append')
            opstack.push(token)
            print(token, 'opstack에 push')
            
        elif token in '+-/*^':
            filtered_prec = {k:v for k, v in prec.items() if 0 < v <= prec['+']}
            print('필터 딕셔너리: ', filtered_prec)
            while True:
                if len(opstack) != 0 and opstack.top() in filtered_prec.keys():
                    outstack.append(opstack.pop())
                    print('opstack pop한 것을 Outstack으로 append')
                else:
                    opstack.push(token)
                    print(token, 'opstack으로 push')
                    break
        else: # operand일 때
            outstack.append(token)
            print(token, '상수라서 outstack으로 바로 append')

    while True:
        if len(opstack) == 0:
            break
        elif opstack.top() in ['(', ')']:
            opstack.pop()
            print('괄호여서 opstack에서 pop 처리')
        else:
            outstack.append(opstack.pop())
            print('opstack에 남은 연산자를 push')
				
    return " ".join(outstack)

	
infix_expr = input()
postfix_expr = infix_to_postfix(infix_expr)
print('\n결과: ', postfix_expr)
```

## 4. 최종 제출본

```python
'''
Infix to postfix
'''


class Stack:
    def __init__(self):
        self.items = []

    def push(self, val):
        self.items.append(val)

    def pop(self):
        try:
            return self.items.pop()
        except IndexError:
            print("Stack is empty")

    def top(self):
        try:
            return self.items[-1]
        except IndexError:
            print("Stack is empty")

    def __len__(self):
        return len(self.items)

    def isEmpty(self):
        return self.__len__() == 0


def infix_to_postfix(infix):
    opstack = Stack()
    outstack = []
    token_list = infix.split()
		
    # 연산자의 우선순위 설정
    prec = {}
    prec['('] = 0
    prec['+'] = 1
    prec['-'] = 1
    prec['*'] = 2
    prec['/'] = 2
    prec['^'] = 3

    for token in token_list:
        if token == '(':
            opstack.push(token)

        elif token == ')':
            while True:
                if opstack.top() != '(':
                    outstack.append(opstack.pop())
                else:
                    opstack.pop()
                    break
            
        elif token in '+-/*^':
            filtered_prec = {k:v for k, v in prec.items() if prec[token] <= v}
            while True:
                if len(opstack) != 0 and opstack.top() in filtered_prec.keys():
                    outstack.append(opstack.pop())
                else:
                    opstack.push(token)
                    break
        else: # operand일 때
            outstack.append(token)

    while True:
        if len(opstack) == 0:
            break
        else:
            outstack.append(opstack.pop())
				
    return " ".join(outstack)

	
infix_expr = input()
postfix_expr = infix_to_postfix(infix_expr)
print(postfix_expr)
```
