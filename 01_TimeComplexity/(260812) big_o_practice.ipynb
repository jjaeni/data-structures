### Big-O 계산연습

#### 1. $n$

```python
def doSomething(A, n):
    for i in range(0, n/2): # O(n)
        c = A[i] # O(1)
        A[i] = A[n-1-i] # O(1)
        A[n-1-i] = c # O(1)
```

#### 2. $n^2 \ log_2^n$

```python
def doSomething(n):
    count = 0
    for i in range(0, n): # O(n)
        for j in range(0, n): # O(n)
            k=1 # O(1)
            while k<n: # O(log n) >> 2^k < n
                count+=1 # O(1)
                k = k*2 # O(1)
    return count
```

#### 3. $\sqrt n$

```python
def doSomething(n):
    count = 0 # O(1)
    k = 1 # O(1)
    while k*k <= n: # O(sqrt n)
        count+=1 # O(1)
        k+=1 # O(1)
    return count
```

#### 4. $log_2^n$

```python
def doSomething(n):
    k = 1 # O(1)
    count = 0 # O(1)
    while k < n: # O(log n)
        k*=2 # O(1)
        count+=1 # O(1)
    return count
```
