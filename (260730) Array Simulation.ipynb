```python
A = [2, 4, 0, 5]
print(f'A 처음 상태: {A} \n')

A.append(6)
print(f'A append 하고나면?: {A} \n')

A.pop()
print(f'A에서 pop 하고 나면?: {A} \n')

A.append(6)
A.pop(1)
print(f'A에서 pop을 순서 지정하면?: {A} \n')

A.insert(1, 4)
print(f'A 1번째 자리 빼준걸 다시 복구하면?: {A} \n')

A.remove(5)
print(f'A에서 5인 원소를 빼주라고 시키면?: {A} \n')

A.insert(1, 0)
print(A)
print(f'A의 0 원소의 인덱스가 궁금하면?: {A.index(0)}') # 가장 첫번째로 나오는 인덱스 번호만 리턴하게 됨

print(f'A의 0 원소가 몇 개인지 궁금하면?: {A.count(0)} \n')
```

```python
import sys

A = []
print(f'empty 리스트 사이즈: {sys.getsizeof(A)}')

A.append(18)
print(f'A에 18을 추가했을 때의 리스트 사이즈: {sys.getsizeof(A)}')
```
