## 1. Dequeue 로직 테스트

```python
lst = [1, 2, 3, 4, 5]
bucket = []
bucket.append(6)

for i in lst:
    bucket.append(i)
lst = bucket

print(lst)

lst.pop(0)
print(lst)
```
