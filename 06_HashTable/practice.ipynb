#### [3] find_slot

```python
def find_slot(key):
    i = hash_function(key)
    start = 1

    while keys[i] != None and keys[i] != key:
        i = (i+1)%size

        if (i == start):
            return 'FULL'

    return i
```

#### [4] set

```python
def set(key, value = None):
    i = find_slot(key)

    if i == 'FULL':
        return None
    else:
        if keys[i] == key:
              values[i] = value

        elif keys[i] == None:
            keys[i] = key
            values[i] = value

        return key
```

#### [5] search

```python
def search(key):
    i = find_slot(key)

    if keys[i] == key:
        return key
    else:
        return None
```
