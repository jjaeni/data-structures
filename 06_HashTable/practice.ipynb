#### [1] init

```python
def init():
    size = 10
    keys = [None] * 10
    values = [None] * 10
    return keys, values, size

keys, values, size = init()
print(keys)
print(values)
```

#### [2] hash function

```python
# key가 1일 때
key = 11
value = '떵개떵'

def hash_function(key):
    i = key % size
    return i
```

#### [3] find_slot

```python
def find_slot(key):
    i = hash_function(key)
    start = 1

    while keys[i] != None and keys[i] != key:
        i = (i+1)%size

        if (i == start):
            return 'FULL'

    return i
```

#### [4] set

```python
def set(key, value = None):
    i = find_slot(key)

    if i == 'FULL':
        return None
    else:
        if keys[i] == key:
              values[i] = value

        elif keys[i] == None:
            keys[i] = key
            values[i] = value

        return key
```

#### [5] search

```python
def search(key):
    i = find_slot(key)

    if keys[i] == key:
        return key
    else:
        return None
```

#### [6] remove

```python
def remove(key):
    i = find_slot(key)
    j = i

    if keys[i] != key:
        return None
    
    while True:
        keys[i] = None
        values[i] = None

        j = (j+1)%size
        
        if keys[j] == None:
            return key
        
        k = hash_function(keys[j])
        
        print(f'i: {i} k: {k} j: {j}')
        if not (i==k or i<=k<j or j<i<k or k<i<j):
            print('순서가 안 맞아요')
            break
        

        keys[i] = keys[j]
        values[i] = values[j]
        i = j
```
